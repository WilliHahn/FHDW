```python
import numpy as np
import matplotlib.pyplot as plt
from math import pi, gamma

# Volumen der n-dimensionalen Einheitskugel
def unit_ball_volume(n):
    return pi**(n/2) / gamma(n/2 + 1)

# Dimensionen
dims = np.arange(1, 23)

# Volumina berechnen
volumes = [unit_ball_volume(n) for n in dims]

# Plot
plt.figure(figsize=(8, 5))
plt.plot(dims, volumes, marker='o', linewidth=2)

plt.xlabel("N dimensions")
plt.ylabel("Volume")
plt.title("Volume of the Unit Ball")
plt.grid(True, alpha=0.3)

plt.xlim(1, 22)

plt.tight_layout()
plt.show()
```
