```python
import os
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_openml
from IPython.display import display, HTML

# -------------------------------
# 1. Daten laden oder lokal lesen
# -------------------------------
file_path = "mnist_test_data.npz"

if os.path.exists(file_path):
    print("Lade Daten lokal...")
    data = np.load(file_path)
    X = data["X"]
    y = data["y"]
else:
    print("Lade MNIST von OpenML...")
    mnist = fetch_openml("mnist_784", version=1, as_frame=False)
    
    X = mnist.data.astype(np.uint8)
    y = mnist.target.astype(int)
    
    # Speichern
    np.savez(file_path, X=X, y=y)
    print("Daten lokal gespeichert.")

# -------------------------------
# 2. Erste '7' finden
# -------------------------------
idx_7 = np.where(y == 7)[0][0]
image = X[idx_7]

# -------------------------------
# 3. 28x28 Matrix darstellen
# -------------------------------
image_matrix = image.reshape(28, 28)
df_matrix = pd.DataFrame(image_matrix)

# Scrollbarer Container (horizontal + vertikal)
def display_scroll(df, width=500, height=300):
    display(HTML(
        f'<div style="width:{width}px; height:{height}px; overflow:auto; border:1px solid black;">'
        + df.to_html()
        + '</div>'
    ))

print("28x28 Grauwertmatrix:")
display_scroll(df_matrix)

# -------------------------------
# 4. Flatten zu 1D-Vektor
# -------------------------------
vector = image.flatten()

# Einzeiliges DataFrame (keine Umbrüche)
df_vector = pd.DataFrame([vector])

# -------------------------------
# 5. Farbmarkierung Vielfache von 28
# -------------------------------
def highlight_multiples(val):
    return ''

def highlight_positions(df):
    styles = pd.DataFrame('', index=df.index, columns=df.columns)
    for col in df.columns:
        if col % 28 == 0:
            styles.iloc[0, col] = 'background-color: lightgreen'
    return styles

styled_vector = df_vector.style.apply(highlight_positions, axis=None)

# -------------------------------
# 6. Anzeige (nur horizontal scroll)
# -------------------------------
print("Flattened 1D Vektor (Vielfache von 28 markiert):")

html_vector = styled_vector.to_html()

display(HTML(
    '<div style="width:100%; overflow-x:auto; white-space:nowrap; border:1px solid black;">'
    + html_vector +
    '</div>'
))
```
