```python
import pandas as pd
import sweetviz as sv

### Daten einlesen
path = 'c:/myBox/Projekte/FHDW/Kurs_DAML/Vorlesung_v10/python/AmesHousing.csv' # für lokale Dateien
#path = 'https://github.com/WilliHahn/FHDW/raw/main/AmesHousing.csv'

df = pd.read_csv(path, sep=',') 

### Analyse ausführen und die Zielvariable SalePrice benennen
#
report_target = sv.analyze([df,'All'], target_feat='SalePrice' )
#report_target.show_html(layout='vertical', scale=0.8, open_browser=True) #erzeugt HTML Datei, nicht auf Colab
#report_target.show_html('Ames_EDA_Sweetviz.html', open_browser=True)
#IPython.display.HTML("Ames_EDA_Sweetviz.html")
report_target.show_notebook() # if working in colab
```
